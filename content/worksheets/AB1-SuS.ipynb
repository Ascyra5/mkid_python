### Mini-Projekt: Dein Ermittlungs-Taschenrechner
Programmiere einen kleinen Taschenrechner für die Ermittlung. Der Code sieht erstmal kompliziert aus, aber so schlimm ist er gar nicht. Kennst du das englische Wort if? Das bedeutet **wenn**.  Wir sagen hier also sowas wie "Wenn das Zeichen `+` ist, dann rechne die beiden Zahlen zusammen." 
Wenn wir zwei zaheln `a` und `b` haben, wie kann man die in Python miteinander addieren?
Ersetze im untenstehenden Code jeweils `pass` durch die passende Rechenoperation. Führe den Code danach aus und teste ihn mit eigenen Zahlen.




```python
def berechne(a, b, operation):

    if operation == "+":
        return a+b  # Ersetze pass durch deine Rechnung

    elif operation == "-":
        return a-b

    elif operation == "*":
        return a*b

    elif operation == "/":
        return a/b
```

```python
# Probiere hier deinen Rechner aus. Bsp.: gib ein 'berechne(zahl1, zahl2, "+")' ein.
```

```python
a = float(input("Erste Zahl: "))
b = float(input("Zweite Zahl: "))
operation = input("Rechenzeichen (+, -, *, /): ")

ergebnis = berechne(a, b, operation)

print("Dein Ergebnis ist:", ergebnis)
```
